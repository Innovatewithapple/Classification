```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#model + preprocessing
from sklearn.model_selection import train_test_split,GridSearchCV,cross_validate
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

#model
from sklearn.linear_model import LogisticRegression

#Evaluation
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
df = pd.read_csv('/content/adult.csv',skiprows=1)
df.head(5)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.info()
```

```python
x = df.drop(columns='incomes')
y = df['incomes']
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
scalingColumns = x.columns
preprocessor = ColumnTransformer([
    ('nums',StandardScaler(),scalingColumns)
])
```

```python
#Logistic
logisticPipeline = Pipeline([
    ('preprocessor',preprocessor),
    ('model',LogisticRegression())
])
```

```python
logisticPipeline.fit(x_train,y_train)
```

```python
cross_score = cross_validate(logisticPipeline,x_train,y_train,cv=5,scoring=["accuracy", "precision", "recall", "f1"])
cross_score
```
